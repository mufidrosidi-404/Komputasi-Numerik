### Cell 1 · Siapkan perkakas

```python
import random
import numpy as np
print("Perkakas siap.")
```

### Cell 2 · Masukkan SPL yang mau diselesaikan

```python
koefisien_a = [
    [ 3,  5, -2],
    [ 6, -1,  4],
    [-2,  7,  3],
]
ruas_kanan_b = [-17, 55, -8]
nama_peubah = ["p", "q", "r"]

print("SPL yang akan diselesaikan:")
for baris, kanan in zip(koefisien_a, ruas_kanan_b):
    suku = [f"{k:+d}{nama}" for k, nama in zip(baris, nama_peubah)]
    print("  ", " ".join(suku), "=", kanan)
```

### Cell 3 · Fungsi pencetak papan

```python
def cetak_papan(papan, judul):
    print(f"\n{judul}")
    for baris in papan:
        kiri = "  ".join(f"{round(v, 10) + 0.0:9.4f}" for v in baris[:-1])
        print(f"  [ {kiri} | {round(baris[-1], 10) + 0.0:9.4f} ]")
```

### Cell 4 · Alat bantu operasi baris

```python
def rakit_papan(koef, kanan):
    return [list(map(float, baris)) + [float(k)] for baris, k in zip(koef, kanan)]

def tukar_dua_baris(papan, i, j):
    papan[i], papan[j] = papan[j], papan[i]

def cari_baris_terkuat(papan, kolom, mulai):
    terbaik = mulai
    for r in range(mulai + 1, len(papan)):
        if abs(papan[r][kolom]) > abs(papan[terbaik][kolom]):
            terbaik = r
    return terbaik

def jadikan_satu(papan, r, kolom):
    pembagi = papan[r][kolom]
    papan[r] = [v / pembagi for v in papan[r]]

def kosongkan_kolom(papan, r_acuan, kolom):
    for r in range(len(papan)):
        if r == r_acuan:
            continue
        faktor = papan[r][kolom]
        papan[r] = [v - faktor * a for v, a in zip(papan[r], papan[r_acuan])]
```

### Cell 5 · Mesin Gauss-Jordan

```python
BATAS_NOL = 1e-12

def selesaikan_spl(koef, kanan, tampil_langkah=True):
    papan = rakit_papan(koef, kanan)
    n = len(papan)
    if tampil_langkah:
        cetak_papan(papan, "Papan awal")
    for kol in range(n):
        r_kuat = cari_baris_terkuat(papan, kol, kol)
        if abs(papan[r_kuat][kol]) < BATAS_NOL:
            raise ValueError(f"Kolom ke-{kol + 1} tidak punya pivot: SPL tidak punya solusi tunggal.")
        if r_kuat != kol:
            tukar_dua_baris(papan, kol, r_kuat)
            if tampil_langkah:
                cetak_papan(papan, f"Tukar baris {kol + 1} dengan baris {r_kuat + 1}")
        jadikan_satu(papan, kol, kol)
        if tampil_langkah:
            cetak_papan(papan, f"Baris {kol + 1} dibagi pivot (pivot jadi 1)")
        kosongkan_kolom(papan, kol, kol)
        if tampil_langkah:
            cetak_papan(papan, f"Kolom {kol + 1} dikosongkan di baris lain")
    return [papan[r][-1] for r in range(n)]
```

### Cell 6 · Jalankan dan lihat hasilnya

```python
hasil = selesaikan_spl(koefisien_a, ruas_kanan_b)
print("\nJawaban akhir:")
for nama, nilai in zip(nama_peubah, hasil):
    print(f"  {nama} = {nilai:.4f}")
```

### Cell 7 · Cek jawaban dengan substitusi

```python
sisa = []
for baris, kanan in zip(koefisien_a, ruas_kanan_b):
    kiri = sum(k * v for k, v in zip(baris, hasil))
    sisa.append(kiri - kanan)
    print(f"ruas kiri = {kiri:9.5f} | ruas kanan = {kanan:4d} | selisih = {kiri - kanan:.2e}")
print("Semua persamaan terpenuhi?", all(abs(s) < 1e-9 for s in sisa))
```

### Cell 8 · Uji kasus tanpa solusi tunggal

```python
koef_kembar = [[1, 2, 3], [2, 4, 6], [1, 0, 1]]
kanan_kembar = [4, 8, 2]
try:
    selesaikan_spl(koef_kembar, kanan_kembar, tampil_langkah=False)
except ValueError as galat:
    print("Program berhenti dengan pesan:", galat)
```

### Cell 9 · Uji acak: bandingkan dengan NumPy

```python
random.seed(20251110)

def buat_spl_acak():
    while True:
        a = [[random.randint(-9, 9) for _ in range(3)] for _ in range(3)]
        if abs(np.linalg.det(np.array(a))) > 1e-6:
            return a

for uji_ke in range(1, 4):
    a = buat_spl_acak()
    kunci = [random.randint(-9, 9) for _ in range(3)]
    b = [sum(k * v for k, v in zip(baris, kunci)) for baris in a]
    tebakan = selesaikan_spl(a, b, tampil_langkah=False)
    cocok = np.allclose(tebakan, np.linalg.solve(np.array(a, float), np.array(b, float)))
    print(f"Uji {uji_ke}: kunci = {kunci} | tebakan = {[round(t, 4) + 0.0 for t in tebakan]} | cocok dengan NumPy: {cocok}")
```
